````python
# готовые инструменты для проверок — писать их не нужно, просто запустите ячейку
import inspect

_KIND_NAMES = {
    inspect.Parameter.POSITIONAL_ONLY: "только по позиции",
    inspect.Parameter.POSITIONAL_OR_KEYWORD: "обычный",
    inspect.Parameter.VAR_POSITIONAL: "*args",
    inspect.Parameter.KEYWORD_ONLY: "только по имени",
    inspect.Parameter.VAR_KEYWORD: "**kwargs",
}


def kinds(func):
    # словарь «имя параметра -> вид»: как параметр можно передавать
    return {name: _KIND_NAMES[p.kind] for name, p in inspect.signature(func).parameters.items()}


def raises(exc, func, *args, **kwargs):
    # True, если вызов func(*args, **kwargs) бросил исключение exc
    try:
        func(*args, **kwargs)
    except exc:
        return True
    return False


print("инструменты для проверок загружены")
````


````python
DISCOUNT = 10      # обычная скидка, в процентах


def final_price(price: int) -> int:
    """Цена со скидкой; от 1000 ₽ скидка на 5 процентных пунктов больше."""
    # исправьте функцию
    if price >= 1000:
        DISCOUNT = DISCOUNT + 5
    return price * (100 - DISCOUNT) // 100


assert final_price(500) == 450
assert final_price(999) == 899
assert final_price(1000) == 850
assert final_price(2000) == 1700
assert final_price(0) == 0
assert DISCOUNT == 10, "глобальная скидка не должна меняться"
assert final_price(1000) == 850, "повторный вызов должен давать то же самое"

DISCOUNT = 20
new_prices = (final_price(500), final_price(1000))
DISCOUNT = 10
assert new_prices == (400, 750), "скидку нужно брать из DISCOUNT в момент вызова"
print("Задача 4: ок")
````
