### Задача 1.1. Сумма цифр

Верните сумму цифр целого числа. Для отрицательного числа знак не учитывается.

```
digits_sum(1234)  ->  10
digits_sum(-907)  ->  16
digits_sum(0)     ->  0
```

*Подсказка:* число превращается в строку функцией `str`, а каждый символ обратно в цифру — `int`.

````python
def digits_sum(n: int) -> int:
    """Сумма цифр числа n."""
    # ваш код здесь
    sum=0
    for i in str(n):
      if i.isdigit():
        sum+=int(i)
    return sum

    pass


assert digits_sum(0) == 0
assert digits_sum(9) == 9
assert digits_sum(1234) == 10
assert digits_sum(-907) == 16
print("Задача 1.1: ок")
````

```

to_base(255, 16)  ->  "FF"
```

*Подсказка:* деление с остатком даёт цифры справа налево, значит в конце их нужно перевернуть.
Встроенными `bin`, `oct`, `hex` пользоваться нельзя.

````python
def to_base(n: int, b: int) -> str:
    if n == 0:
        return "0"
    digits = "0123456789ABCDEF"
    res = []
    while n > 0:
        res.append(digits[n % b])
        n //= b
        #print(n)
    #print(res)
    return "".join(reversed(res))
    # ваш код здесь
    pass


assert to_base(0, 2) == "0"
assert to_base(5, 2) == "101"
assert to_base(255, 16) == "FF"
assert to_base(255, 8) == "377"
assert to_base(4095, 16) == "FFF"
print("Задача 1.2: ок")
````

### Задача 1.4. Лишние пробелы

Уберите пробелы в начале и в конце строки, а каждую группу пробельных символов
внутри замените одним пробелом.

```
normalize_spaces("  привет   мир  ")  ->  "привет мир"
normalize_spaces("   ")               ->  ""
```

*Подсказка:* `split()` без аргументов сам разбивает по любым пробельным символам и выбрасывает пустые куски.

````python
def normalize_spaces(s: str) -> str:
    """Строка без лишних пробелов."""
    sl=[]
    le=False
    for i in s:
        za=(' ','\n','\t')

        if i not in za :
            sl.append(i)
            le=True
        else:
            if le:
                sl.append(" ")
                le =False

    if len(sl):
        if sl[len(sl)-1]==' ':
            sl.pop()
    return "".join(sl)

assert normalize_spaces("  привет   мир  ") == "привет мир"
assert normalize_spaces("одно") == "одно"
assert normalize_spaces("   ") == ""
assert normalize_spaces("a\tb\nc") == "a b c"
print("Задача 1.4: ок")
````

````python
def longest_word(text: str) -> str:
    """Самое длинное слово текста."""
    # ваш код здесь
    pun = ",.!?;:-—"
    ctext = text
    for c in pun:
      ctext = ctext.replace(c, " ")
      ctx=ctext.split()



    st=sorted(ctx,key=lambda x: (-len(x),ctx.index(x)))
    print(st)
    if st:
      return st[0]
    else:
      return ''

    pass


assert longest_word("мама мыла раму") == "мама"
assert longest_word("мама мыла грязную раму") == "грязную"
assert longest_word("а бб ввв, гггг!") == "гггг"
assert longest_word("одинаково слово") == "одинаково"
assert longest_word("") == ""
print("Задача 1.6: ок")
````

### Задача 1.7. Разряды числа

Запишите целое число, разделив цифры на группы по три справа налево обычным пробелом.

```
money(1234567)  ->  "1 234 567"
money(1000)     ->  "1 000"
money(999)      ->  "999"
money(-45678)   ->  "-45 678"
```

*Подсказка:* удобно резать строку срезами с конца. Форматом `f"{n:,}"` пользоваться нельзя.

````python
def money(n: int) -> str:
    """Число с разделителями разрядов."""
    # ваш код здес
    cnt = 0
    ans = []
    s = str(abs(n))
    st = len(s) % 3 or 3
    for idx, i in enumerate(s):
        ans.append(i)
        cnt += 1
        #print(ans)
        if cnt == st and idx < len(s) - 1:
            ans.append(' ')
            cnt = 0
            st = 3
    if n < 0:
        return '-' + ''.join(ans)
    return ''.join(ans)




    pass


assert money(0) == "0"
assert money(999) == "999"
assert money(1000) == "1 000"
assert money(1234567) == "1 234 567"
assert money(-45678) == "-45 678"
print("Задача 1.7: ок")
````

### Задача 1.8. Римские цифры

```
roman(4)     ->  "IV"
roman(14)    ->  "XIV"
roman(1987)  ->  "MCMLXXXVII"
roman(3999)  ->  "MMMCMXCIX"
```

*Подсказка:* заведите список пар «значение — обозначение» от большего к меньшему, включая
900 → CM, 400 → CD, 90 → XC, 40 → XL, 9 → IX, 4 → IV. Тогда достаточно жадно вычитать.

````python
def roman(n: int) -> str:
    """Римская запись числа n."""
    # ваш код здесь
    vals = [
        (1000, "M"), (900, "CM"), (500, "D"), (400, "CD"),
        (100, "C"), (90, "XC"), (50, "L"), (40, "XL"),
        (10, "X"), (9, "IX"), (5, "V"), (4, "IV"), (1, "I")
    ]
    res = []
    for v, sym in vals:
        while n >= v:
            res.append(sym)
            n -= v
    return "".join(res)
    pass


assert roman(1) == "I"
assert roman(4) == "IV"
assert roman(9) == "IX"
assert roman(14) == "XIV"
assert roman(1987) == "MCMLXXXVII"
assert roman(3999) == "MMMCMXCIX"
print("Задача 1.8: ок")
````

````python
def rle_decode(code: str) -> str:
    """Восстановленная строка."""
    # ваш код здесь

    res = []
    i = 0
    while i < len(code):
        c = code[i]
        i += 1

        num = ""

        while i < len(code) and code[i].isdigit():
            num += code[i]

            i += 1
        cnt = int(num) if num else 1


        res.append(c * cnt)
    return "".join(res)


    pass


assert rle_decode("a3bc4d") == "aaabccccd"
assert rle_decode("abc") == "abc"
assert rle_decode("") == ""
assert rle_decode("z12") == "z" * 12
print("Задача 3.2: ок")
````

````python
def rotated(matrix: list) -> list:
    n = len(matrix)


    res = []
    for i in range(n):
        row = []

        for r in range(n - 1, -1, -1):

            row.append(matrix[r][i])
        res.append(row)


    return res



assert rotated([[1, 2], [3, 4]]) == [[3, 1], [4, 2]]
assert rotated([[1]]) == [[1]]
assert rotated([[1, 2, 3], [4, 5, 6], [7, 8, 9]]) == [[7, 4, 1], [8, 5, 2], [9, 6, 3]]
m = [[1, 2], [3, 4]]
rotated(m)
assert m == [[1, 2], [3, 4]], "исходную матрицу менять нельзя"
print("Задача 3.4: ок")
````
